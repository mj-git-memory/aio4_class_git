### 1. 함수(Function)란 무엇인가요?

함수는 특정 작업을 수행하는 **코드의 상자**라고 생각하면 쉽습니다.

* **입력(Parameter/Argument):** 함수에 전달하는 값입니다. (없을 수도 있습니다)
* **처리(Body):** 함수 내부에서 수행할 코드들입니다.
* **출력(Return):** 처리가 끝난 후 돌려주는 결과값입니다. (없을 수도 있습니다)

#### 함수의 기본 구조:
```python
def 함수이름(매개변수):
    # 실행할 코드
    return 결과값
```

````python
# [예시] 인사말을 만들어주는 간단한 함수
def greet_user(name):
    """
    이름을 입력받아 환영 메시지를 반환하는 함수입니다.
    """
    message = f"안녕하세요, {name}님! 파이썬 공부를 환영합니다."
    return message

# 함수 사용(호출)하기
result = greet_user("철수")
print(result)

result2 = greet_user("영희")
print(result2)
````

````python
# 1. 리스트 생성 (Create)
# 다양한 데이터 타입을 담거나 빈 리스트를 만들 수 있습니다.
my_list = [10, 20, 30, 40]
print("처음 생성된 리스트:", my_list)

# 새로운 요소 추가 (맨 뒤에 추가)
my_list.append(50)
print("요소 추가 후 (append):", my_list)

# 특정 위치에 요소 삽입 (인덱스 1 자리에 15 삽입)
my_list.insert(1, 15)
print("요소 삽입 후 (insert):", my_list)

# 2. 리스트 변경 (Update)
# 인덱스를 활용하여 특정 위치의 값을 직접 변경합니다.
my_list[2] = 200  # 인덱스 2의 값을 200으로 변경 (기존 20)
print("요소 변경 후:", my_list)

# 3. 리스트 삭제 (Delete)
# (1) del 키워드를 이용해 특정 인덱스의 요소를 삭제
del my_list[1]  # 인덱스 1의 값(15) 삭제
print("인덱스 1 삭제 후 (del):", my_list)

# (2) remove() 메서드를 이용해 특정 값을 직접 삭제
my_list.remove(200)  # 리스트에서 200이라는 값을 찾아 삭제
print("값 200 삭제 후 (remove):", my_list)

# (3) pop() 메서드로 맨 마지막 값을 꺼내며 삭제
popped_value = my_list.pop()
print(f"꺼내진 값: {popped_value}, pop 후 리스트:", my_list)
````

### 1. 리스트 슬라이싱 (Slicing)
리스트의 일부 구간을 잘라내어 새로운 리스트를 만드는 방법입니다.
* **형식:** `list[start:end:step]` (start는 포함, end는 미포함)

### 2. 리스트 정렬 (Sorting)
리스트의 요소를 크기 순서대로 정렬하는 방법입니다.
* `sort()`: 원본 리스트 자체를 정렬합니다.
* `sorted()`: 원본은 그대로 두고 정렬된 새로운 리스트를 반환합니다.

### 3. 리스트 컴프리헨션 (List Comprehension)
한 줄의 짧고 직관적인 코드로 새로운 리스트를 생성하는 아주 강력한 파이썬 문법입니다.
* **형식:** `[표현식 for 항목 in 반복가능객체 if 조건]`

````python
# 1. 슬라이싱 (Slicing) 예제
numbers = [0, 10, 20, 30, 40, 50, 60]
print("원본 리스트:", numbers)
print("인덱스 1부터 4까지:", numbers[1:5])  # [10, 20, 30, 40]
print("처음부터 3까지:", numbers[:4])       # [0, 10, 20, 30]
print("간격 2로 건너뛰며:", numbers[::2])     # [0, 20, 40, 60]
print("리스트 뒤집기:", numbers[::-1])       # [60, 50, 40, 30, 20, 10, 0]
print("-" * 40)

# 2. 정렬 (Sorting) 예제
unsorted_list = [30, 10, 50, 20, 40]

# sorted()는 원본을 건드리지 않고 정렬본을 만듦
sorted_new = sorted(unsorted_list)
print("원본 리스트:", unsorted_list)
print("sorted() 적용 결과:", sorted_new)

# sort()는 원본 자체를 정렬함 (역정렬: reverse=True)
unsorted_list.sort(reverse=True)
print("sort(reverse=True) 후 원본:", unsorted_list)
print("-" * 40)

# 3. 리스트 컴프리헨션 (List Comprehension) 예제
# [예제 1] 0부터 9까지 숫자 중 짝수만 골라 제곱한 리스트 만들기
even_squares = [x**2 for x in range(10) if x % 2 == 0]
print("짝수 제곱 리스트:", even_squares)

# [예제 2] 리스트의 모든 문자열을 대문자로 바꾸기
words = ["apple", "banana", "cherry"]
upper_words = [word.upper() for word in words]
print("대문자 변환 리스트:", upper_words)
````

### 리스트에서 모든 특정 값(예: 200)을 한 번에 지우는 방법

#### 방법 1: 리스트 컴프리헨션 사용 (가장 추천하는 파이썬다운 방식)
지우고 싶은 값(200)이 **아닌** 요소들만 필터링하여 새로운 리스트를 만듭니다.

#### 방법 2: `while` 루프 사용
리스트 안에 지우고자 하는 값이 있는 동안 계속해서 `remove()`를 반복 호출합니다.

````python
# 중복된 200이 들어있는 리스트 준비
test_list = [10, 200, 30, 200, 40, 200, 50]
print("원본 리스트:", test_list)

# 1. 리스트 컴프리헨션 방식 (새로운 리스트를 생성)
filtered_list = [x for x in test_list if x != 200]
print("방법 1 (컴프리헨션) 결과:", filtered_list)

# 2. while 루프 방식 (원본 리스트를 직접 수정)
while 200 in test_list:
    test_list.remove(200)
print("방법 2 (while 루프) 결과:", test_list)
````

### 중첩 조건문 (Nested Conditional Statement)

조건문 내부의 실행 코드 블록 안에 **또 다른 `if-elif-else` 조건문을 넣는 구조**를 의미합니다. 들여쓰기(Indentation) 규칙만 잘 지키면 다중 레벨로 조건문 안에 조건문을 깊게 설계할 수 있습니다.

```python
if 바깥쪽_조건식_1:
    # 바깥쪽 조건 1이 만족할 때 실행할 코드
    if 안쪽_조건식_A:
        # 안쪽 조건 A가 만족할 때 실행
    else:
        # 안쪽 조건 A가 불만족할 때 실행
else:
    # 바깥쪽 조건 1이 만족하지 않을 때 실행할 코드
```

````python
# [예제] 쇼핑몰 회원 여부 및 구매 금액별 배송비 계산
is_member = False      # 회원 여부 (True: 회원, False: 비회원)
purchase_amount = 45000  # 구매 금액 (원)

if is_member:
    print("회원 로그인 상태입니다.")
    # 회원 전용 중첩 조건문
    if purchase_amount >= 30000:
        shipping_fee = 0
        print("회원 혜택 적용: 3만원 이상 무료 배송!")
    else:
        shipping_fee = 3000
        print("3만원 미만 구매로 배송비 3,000원이 부과됩니다.")
else:
    print("비회원 상태입니다.")
    # 비회원 전용 중첩 조건문
    if purchase_amount >= 50000:
        shipping_fee = 0
        print("비회원 혜택 적용: 5만원 이상 무료 배송!")
    else:
        shipping_fee = 3500
        print("5만원 미만 구매로 배송비 3,500원이 부과됩니다.")

print(f"최종 배송비: {shipping_fee}원")
````

### 반복문(Loop)을 활용해 특정 값 모두 제거하기

#### 방법 1: `for` 루프와 새로운 리스트 생성 (가장 안전한 방법)
원본 리스트의 요소를 하나씩 꺼내어 확인한 뒤, '사과'가 아니면 새로운 리스트에 차곡차곡 쌓는 방식입니다. 원본 데이터를 유지하면서 작업할 수 있어 안전합니다.

#### 방법 2: `while` 루프와 `remove()` 사용 (원본 직접 수정)
리스트 안에 '사과'가 남아있는 동안 계속해서 `remove('사과')`를 반복 실행합니다.

````python
# 1. for 루프를 사용하는 방법
fruits = ['사과', '바나나', '오렌지', '사과', '키위', '사과', '체리', '바나나']
clean_fruits = []

for fruit in fruits:
    if fruit != '사과':
        clean_fruits.append(fruit)

print("방법 1 (for 루프) 결과:", clean_fruits)


# 2. while 루프를 사용하는 방법
fruits_copy = ['사과', '바나나', '오렌지', '사과', '키위', '사과', '체리', '바나나']

while '사과' in fruits_copy:
    fruits_copy.remove('사과')

print("방법 2 (while 루프) 결과:", fruits_copy)
````

### input()으로 입력받은 값을 숫자로 변환하기

`input()` 함수가 돌려주는 값은 언제나 문자열입니다. 따라서 수학적인 계산을 하려면 `int()` 나 `float()` 함수를 사용해 숫자로 타입을 직접 바꾸어 주어야(형변환) 합니다.

````python
# 1. 정수로 변환하기 (int)
print("--- 1. 정수형 변환 예시 ---")
num_str1 = input("첫 번째 숫자를 입력하세요: ")
num_str2 = input("두 번째 숫자를 입력하세요: ")

# int()를 사용해 문자열을 정수로 변환합니다.
value1 = int(num_str1)
value2 = int(num_str2)

print(f"두 수의 합: {value1 + value2}")
````

### `num_input` 함수 만들기

두 수를 입력받아 정수로 변환한 뒤, 그 합을 반환하는 함수 예시입니다.

````python
def num_input():
    print("--- num_input() 함수가 실행되었습니다 ---")
    num_str1 = input("첫 번째 숫자를 입력하세요: ")
    num_str2 = input("두 번째 숫자를 입력하세요: ")

    # 정수로 변환
    value1 = int(num_str1)
    value2 = int(num_str2)

    # 두 수의 합을 반환(return)
    return value1 + value2

# 함수를 호출해서 결과를 변수에 저장하고 출력해 봅니다.
total_sum = num_input()
print(f"함수가 반환한 두 수의 합: {total_sum}")
````
