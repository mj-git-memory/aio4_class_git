## 3. 리스트, for, range, f-string

원본: `day1-3.py`, 노트북의 짝수 추출 셀

- `scores = []`: 빈 리스트를 만듭니다. 리스트는 여러 값을 저장하며 내용을 변경할 수 있습니다.
- `scores.append(score)`: 리스트 끝에 값을 추가하는 메서드 호출입니다.
- `for i in range(5)`: `i`가 0, 1, 2, 3, 4로 바뀌며 다섯 번 반복합니다.
- `range(시작, 끝)`에서 끝값은 포함되지 않습니다. `range(1, 5)`는 1~4입니다.
- `for value in scores`: 리스트의 값을 하나씩 꺼냅니다.
- `f"{i + 1}번째"`: 문자열의 중괄호 안에 변수나 계산 결과를 넣는 f-string입니다.

````python
# 따라 하기: 1, 2, 3, 4, 5를 차례로 입력하세요.
scores = []
for i in range(5):
    score = int(input(f"{i + 1}번째 숫자 입력: "))
    scores.append(score)

scores_filter = []
for value in scores:
    if value % 2 == 0:
        scores_filter.append(value)

print("입력한 숫자:", scores)
print("짝수:", scores_filter)
# 예상 짝수: [2, 4]
````

## 4. 중첩 반복문과 출력 형식

원본: `day1-4.py`

- 반복문 안에 반복문을 넣으면 바깥 반복 한 번마다 안쪽 반복이 처음부터 끝까지 실행됩니다.
- `print()`는 기본적으로 출력 끝에 줄바꿈을 붙입니다.
- `end="\t"`는 키워드 인자로, 출력 끝을 탭으로 바꿉니다. `\t`는 탭, `\n`은 줄바꿈입니다.
- 인자 없는 `print()`는 줄바꿈만 출력합니다. `print("\n")`는 문자열의 줄바꿈에 기본 줄바꿈까지 더해집니다.
- 원본의 `''' ... '''`는 설명용으로 사용했지만 문법상 여러 줄 문자열입니다. `#` 주석과 구분하세요.

````python
# 따라 하기: 한 행에 2~5단을 출력합니다.
for multiplier in range(1, 4):
    for dan in range(2, 6):
        print(f"{dan} x {multiplier} = {dan * multiplier}", end="	")
    print()
````

## 6. 모듈, 난수, 횟수 누적, break

원본: `day1-5.py`

- `import random`: 파이썬 표준 라이브러리 모듈을 가져옵니다. 별도 설치는 필요 없습니다.
- `random.randint(1, 100)`: 양 끝을 포함해 1~100 중 정수 하나를 선택합니다. 끝값을 제외하는 `range()`와 다릅니다.
- `count += 1`: `count = count + 1`과 같은 효과로 시도 횟수를 늘립니다.
- `for _ in range(10)`: 최대 10회 반복합니다. `_`는 여기서 반복 변수의 값을 사용하지 않겠다는 관례적인 이름입니다.
- `break`: 정답을 맞히면 반복을 즉시 종료합니다.
- 원본의 `while True`는 주석 처리되어 실행되지 않습니다. 실제 동작하는 코드는 `for`를 사용합니다. `while True`를 사용할 때는 종료 조건을 마련해야 합니다.

````python
# 따라 하기: 정답을 42로 고정해 동작을 쉽게 확인합니다.
# 50, 30, 42를 차례로 입력하면 3번째 시도에서 종료됩니다.
num = 42
count = 0
for _ in range(10):
    count += 1
    user_num = int(input(f"{count}번째 숫자 입력: "))
    if user_num > num:
        print("정답보다 크다")
    elif user_num < num:
        print("정답보다 작다")
    else:
        print(f"정답: {num}. 총 시도 횟수: {count}")
        break
````
